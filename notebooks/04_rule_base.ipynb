this file is for define the rule base that will be used in both Mamdani and Sugeno fuzzy systems

each rule follows the format IF [conditions] THEN [output]

```python
# each rule is a tuple: (nkill, nwound, propextent, severity)
# linguistic terms are represented as strings

rules = [
    # low severity cases
    ("Low",    "Low",    "None",         "Low"),
    ("Low",    "Low",    "Minor",        "Low"),
    ("Low",    "Medium", "None",         "Low"),

    # medium severity cases
    ("Low",    "Low",    "Major",        "Medium"),
    ("Medium", "Low",    "None",         "Medium"),
    ("Medium", "Medium", "Minor",        "Medium"),
    ("Low",    "Medium", "Major",        "Medium"),

    # high severity cases
    ("Medium", "Medium", "Major",        "High"),
    ("High",   "Low",    "Minor",        "High"),
    ("High",   "Medium", "None",         "High"),
    ("Medium", "High",   "Major",        "High"),
    ("High",   "High",   "Minor",        "High"),

    # critical severity cases
    ("High",    "Medium",  "Major",        "Critical"),
    ("High",    "High",    "Major",        "Critical"),
    ("Extreme", "High",    "Major",        "Critical"),
    ("Extreme", "Extreme", "Catastrophic", "Critical"),
]

print(f"Total rules: {len(rules)}")
```

```python
import pandas as pd

df_rules = pd.DataFrame(rules, columns=["nkill", "nwound", "propextent", "severity"])
df_rules.index += 1
df_rules.index.name = "Rule"
df_rules
```

```python
# print as natural language

for i, (kill, wound, prop, sev) in enumerate(rules, 1):
    print(f"Rule {i:02d}: IF nkill={kill} AND nwound={wound} AND propextent={prop} THEN severity={sev}")
```

## Summary

We defined 16 fuzzy rules covering all 4 severity levels

The logic follows common sense reasoning about terrorism attack severity:
- attacks with few casualties and no property damage are classified as Low
- attacks with moderate casualties or some damage are classified as Medium
- attacks with high casualties or significant damage are classified as High
- attacks with extreme casualties and major damage are classified as Critical
