this file define the rule base that will be used in both Mamdani and Sugeno fuzzy systems

each rule follows the format IF [conditions] THEN [output]

we now have 5 input variables: nkill, nwound, propextent, attack_encoded, weapon_encoded

```python
import pandas as pd

def assign_severity(k, w, p, a, wp):
    kill_score  = {"Low": 0, "Medium": 1, "High": 2, "Extreme": 3}
    wound_score = {"Low": 0, "Medium": 1, "High": 2, "Extreme": 3}
    prop_score  = {"None": 0, "Minor": 1, "Major": 2, "Catastrophic": 3}
    atk_score   = {"Low": 0, "Medium": 1, "High": 2, "Extreme": 3}
    wpn_score   = {"Low": 0, "Medium": 1, "High": 2, "Extreme": 3}

    score = (
        0.35 * kill_score[k]  +
        0.25 * wound_score[w] +
        0.15 * prop_score[p]  +
        0.15 * atk_score[a]   +
        0.10 * wpn_score[wp]
    )

    if score < 0.75:
        return "Low"
    elif score < 1.5:
        return "Medium"
    elif score < 2.25:
        return "High"
    else:
        return "Critical"

kill_levels = ["Low", "Medium", "High", "Extreme"]
prop_levels = ["None", "Minor", "Major", "Catastrophic"]

# each rule is a tuple:
# (nkill, nwound, propextent, attack_encoded, weapon_encoded, severity)
rules = []
for k in kill_levels:
    for w in kill_levels:
        for p in prop_levels:
            for a in kill_levels:
                for wp in kill_levels:
                    sev = assign_severity(k, w, p, a, wp)
                    rules.append((k, w, p, a, wp, sev))

print(f"Total rules: {len(rules)}")
```

## Rule Table

displaying all rules in a readable format

```python
df_rules = pd.DataFrame(rules, columns=[
    "nkill", "nwound", "propextent", "attack", "weapon", "severity"
])
df_rules.index += 1
df_rules.index.name = "Rule"
df_rules
```

```python
for i, (kill, wound, prop, atk, wpn, sev) in enumerate(rules, 1):
    print(f"Rule {i:02d}: IF nkill={kill} AND nwound={wound} AND propextent={prop} "
          f"AND attack={atk} AND weapon={wpn} THEN severity={sev}")
```

## Summary

we defined 22 fuzzy rules covering all 4 severity levels

The logic follows common sense reasoning about terrorism attack severity:
- attacks with few casualties, no property damage, and low-level weapons are Low
- attacks with moderate casualties or some damage with medium weapons are Medium
- attacks with high casualties or significant damage with powerful weapons are High
- attacks with extreme casualties, major damage, and the most lethal weapons are Critical

adding attack type and weapon type as inputs makes the rules more realistic
because a bombing with explosives is inherently more severe than an unarmed assault even if the immediate casualty count is similar
