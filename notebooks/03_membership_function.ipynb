this file define and visualize membership functions for all input and outut
variables that will be used in both Mamdani and Sugeno fuzzy systems.

**Inputs:**
1. `nkill` number of fatalities
2. `nwound` number of injuries
3. `propextent` property damage (inverted scale 0–3)
4. `attack_encoded` attack type (scale 1–4)
5. `weapon_encoded` weapon type (scale 1–4)

**Output:**
- `severity_index` attack severity (0–100)

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.dpi"] = 120
colors = ["#2ecc71", "#f39c12", "#e74c3c", "#8e44ad"]
```

we use **Triangular** and **Trapezoidal** membership functions:

- **Trapezoidal** for the outermost sets (Low and Extreme/Critical)
  because they have a flat region at the boundary
- **Triangular** for middle sets (Medium, High)
  because they peak at a single point

```python
def trimf(x, a, b, c):
    return np.maximum(0, np.minimum((x - a) / (b - a + 1e-9),
                                     (c - x) / (c - b + 1e-9)))

def trapmf(x, a, b, c, d):
    return np.maximum(0, np.minimum(
        np.minimum((x - a) / (b - a + 1e-9), 1),
        (d - x) / (d - c + 1e-9)
    ))
```

## Input 1: Fatalities (`nkill`)

Range: 0–50

| Linguistic | Type | Parameters |
|---|---|---|
| Low | Trapezoidal | [0, 0, 1, 4] |
| Medium | Triangular | [2, 6, 12] |
| High | Triangular | [6, 15, 30] |
| Extreme | Trapezoidal | [25, 40, 50, 50] |

```python
x_kill = np.linspace(0, 50, 1000)

mf_kill = {
    "Low":     trapmf(x_kill, 0, 0, 1, 4),
    "Medium":  trimf(x_kill, 2, 6, 12),
    "High":    trimf(x_kill, 6, 15, 30),
    "Extreme": trapmf(x_kill, 25, 40, 50, 50),
}

plt.figure(figsize=(10, 4))
for (label, mf), color in zip(mf_kill.items(), colors):
    plt.plot(x_kill, mf, label=label, color=color, linewidth=2)
plt.title("Membership Functions: Fatalities (nkill)")
plt.xlabel("Number of fatalities")
plt.ylabel("Membership degree")
plt.legend()
plt.tight_layout()
plt.show()
```
