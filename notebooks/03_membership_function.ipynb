this file is for define and visualize membership functions for all input and output variables that will be used in both Mamdani and Sugeno fuzzy systems

**Inputs:**
1. `nkill` number of fatalities
2. `nwound` number of injuries
3. `propextent` property damage (inverted scale 0–3)

**Output:**
- `severity_index` attack severity (0–100)

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

plt.rcParams["figure.dpi"] = 120
sns_colors = ["#2ecc71", "#f39c12", "#e74c3c", "#8e44ad"]
```

we use **Triangular** and **Trapezoidal** membership functions:

- **Trapezoidal** for the outermost sets (Low and Extreme/Critical)
  because they have a flat region at the boundary
- **Triangular** for middle sets (Medium, High)
  because they peak at a single point

```python
def trimf(x, a, b, c):
    """Triangular membership function."""
    return np.maximum(0, np.minimum((x - a) / (b - a + 1e-9),
                                     (c - x) / (c - b + 1e-9)))

def trapmf(x, a, b, c, d):
    """Trapezoidal membership function."""
    return np.maximum(0, np.minimum(
        np.minimum((x - a) / (b - a + 1e-9), 1),
        (d - x) / (d - c + 1e-9)
    ))
```

## Input 1: Fatalities (`nkill`)

Range: 0 to 50 (capped at 95th percentile ~10, extended to 50 for coverage)

| Linguistic | Type | Parameters |
|---|---|---|
| Low | Trapezoidal | [0, 0, 1, 4] |
| Medium | Triangular | [2, 6, 12] |
| High | Triangular | [8, 20, 35] |
| Extreme | Trapezoidal | [25, 40, 50, 50] |

```python
x_kill = np.linspace(0, 50, 1000)

mf_kill = {
    "Low":     trapmf(x_kill, 0, 0, 1, 4),
    "Medium":  trimf(x_kill, 2, 6, 12),
    "High": trimf(x_kill, 6, 15, 30),
    "Extreme": trapmf(x_kill, 25, 40, 50, 50),
}

plt.figure(figsize=(10, 4))
for (label, mf), color in zip(mf_kill.items(), sns_colors):
    plt.plot(x_kill, mf, label=label, color=color, linewidth=2)

plt.title("Membership Functions: Fatalities (nkill)")
plt.xlabel("Number of fatalities")
plt.ylabel("Membership degree")
plt.legend()
plt.tight_layout()
plt.show()
```

## Input 2: Injuries (`nwound`)

Range: 0 to 80

| Linguistic | Type | Parameters |
|---|---|---|
| Low | Trapezoidal | [0, 0, 2, 6] |
| Medium | Triangular | [3, 10, 20] |
| High | Triangular | [15, 35, 60] |
| Extreme | Trapezoidal | [45, 65, 80, 80] |

```python
x_wound = np.linspace(0, 80, 1000)

mf_wound = {
    "Low":     trapmf(x_wound, 0, 0, 2, 6),
    "Medium":  trimf(x_wound, 3, 10, 20),
    "High":    trimf(x_wound, 15, 35, 60),
    "Extreme": trapmf(x_wound, 45, 65, 80, 80),
}

plt.figure(figsize=(10, 4))
for (label, mf), color in zip(mf_wound.items(), sns_colors):
    plt.plot(x_wound, mf, label=label, color=color, linewidth=2)

plt.title("Membership Functions: Injuries (nwound)")
plt.xlabel("Number of injuries")
plt.ylabel("Membership degree")
plt.legend()
plt.tight_layout()
plt.show()
```

## Input 3: Property Damage (`propextent` inverted)

Range: 0 to 3 (discrete scale)
- 0 = Unknown / No damage
- 1 = Minor
- 2 = Major
- 3 = Catastrophic

| Linguistic | Type | Parameters |
|---|---|---|
| None | Trapezoidal | [0, 0, 0, 0.5] |
| Minor | Triangular | [0.5, 1, 1.5] |
| Major | Triangular | [1.5, 2, 2.5] |
| Catastrophic | Trapezoidal | [2.5, 3, 3, 3] |

```python
x_prop = np.linspace(0, 3, 1000)

mf_prop = {
    "None":         trapmf(x_prop, 0, 0, 0, 0.5),
    "Minor":        trimf(x_prop, 0.5, 1, 1.5),
    "Major":        trimf(x_prop, 1.5, 2, 2.5),
    "Catastrophic": trapmf(x_prop, 2.5, 3, 3, 3),
}

plt.figure(figsize=(10, 4))
for (label, mf), color in zip(mf_prop.items(), sns_colors):
    plt.plot(x_prop, mf, label=label, color=color, linewidth=2)

plt.title("Membership Functions: Property Damage (propextent)")
plt.xlabel("Damage level (0=none, 3=catastrophic)")
plt.ylabel("Membership degree")
plt.legend()
plt.tight_layout()
plt.show()
```

## Output: Severity Index (0–100)

| Linguistic | Type | Parameters |
|---|---|---|
| Low | Trapezoidal | [0, 0, 15, 30] |
| Medium | Triangular | [20, 40, 55] |
| High | Triangular | [45, 60, 75] |
| Critical | Trapezoidal | [65, 80, 100, 100] |

```python
x_out = np.linspace(0, 100, 1000)

mf_out = {
    "Low":      trapmf(x_out, 0, 0, 15, 30),
    "Medium":   trimf(x_out, 20, 40, 55),
    "High":     trimf(x_out, 45, 60, 75),
    "Critical": trapmf(x_out, 65, 80, 100, 100),
}

plt.figure(figsize=(10, 4))
for (label, mf), color in zip(mf_out.items(), sns_colors):
    plt.plot(x_out, mf, label=label, color=color, linewidth=2)

plt.title("Membership Functions: Severity Index (output)")
plt.xlabel("Severity score (0–100)")
plt.ylabel("Membership degree")
plt.legend()
plt.tight_layout()
plt.show()
```
